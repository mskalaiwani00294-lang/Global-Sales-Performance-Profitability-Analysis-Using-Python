```python
# Outlier Detection using IQR

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(col, "Outliers:", len(outliers))
```

* Units Sold: 0 outliers → Values are within the normal range.
- Unit Price: 0 outliers → No extreme price values detected.
- Unit Cost: 0 outliers → Cost values are consistent.
- Total Revenue: 86 outliers → Some unusual revenue values were detected.
- Total Cost: 107 outliers → Some unusual cost values were detected.
- Total Profit: 55 outliers → Some unusual profit values were detected.
- Profit Margin: 78 outliers → Some unusual profit margin values were detected.

```python
#IQR Methond treat outliers
numerical_columns = [
    "Units Sold",
    "Unit Price",
    "Unit Cost",
    "Total Revenue",
    "Total Cost",
    "Total Profit",
    "Profit Margin"
]

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)

print("Outlier treatment completed successfully!")
```

# Outlier Treatment Interpretation:
- IQR method is used to treat the detected outliers in the dataset.
- Values below the lower limit are replaced with the lower limit.
- Values above the upper limit are replaced with the upper limit using clip().
- This reduces the effect of extreme values without deleting any rows from the dataset.

```python
##What will be the predicted revenue for the next 6 months?



from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Prepare monthly revenue
df["Order Date"] = pd.to_datetime(df["Order Date"])

revenue_col = (
    "Total Revenue" if "Total Revenue" in df.columns else "Sales"
)

monthly_revenue = (
    df.set_index("Order Date")[revenue_col]
    .resample("MS")
    .sum()
    .sort_index()
)

# Remove months before the first and after the last actual order
monthly_revenue = monthly_revenue.loc[
    monthly_revenue.first_valid_index():
    monthly_revenue.last_valid_index()
].asfreq("MS")

if monthly_revenue.isna().any():
    raise ValueError(
        "Missing monthly revenue found. Check missing months "
        "and fill them only if justified by your data."
    )

if len(monthly_revenue) < 36:
    raise ValueError("At least 36 monthly observations are recommended.")

# 2. Split data: 80% training, 20% testing
split = int(len(monthly_revenue) * 0.80)

train = monthly_revenue.iloc[:split]
test = monthly_revenue.iloc[split:]

if len(train) < 24 or len(test) < 6:
    raise ValueError(
        "Need at least 24 training months and 6 testing months."
    )

# 3. Fit forecasting models
models = {}

models["Holt-Winters Additive"] = ExponentialSmoothing(
    train,
    trend="add",
    seasonal="add",
    seasonal_periods=12,
    initialization_method="estimated"
).fit(optimized=True)

models["Holt-Winters Damped"] = ExponentialSmoothing(
    train,
    trend="add",
    damped_trend=True,
    seasonal="add",
    seasonal_periods=12,
    initialization_method="estimated"
).fit(optimized=True)

# 4. Evaluate models on the test period
results = []
predictions = {}

# Seasonal naive: use the previous year's corresponding months
seasonal_naive = pd.Series(
    [train.iloc[-12 + (i % 12)] for i in range(len(test))],
    index=test.index
)

predictions["Seasonal Naive"] = seasonal_naive

for name, model in models.items():
    predictions[name] = model.forecast(len(test))

for name, pred in predictions.items():

    mae = mean_absolute_error(test, pred)
    rmse = np.sqrt(mean_squared_error(test, pred))

    nonzero = test != 0
    mape = (
        np.mean(np.abs(
            (test[nonzero] - pred[nonzero]) / test[nonzero]
        )) * 100
        if nonzero.any() else np.nan
    )

    denominator = np.abs(test).sum()
    wape = (
        np.abs(test - pred).sum() / denominator * 100
        if denominator != 0 else np.nan
    )

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "MAPE (%)": mape,
        "WAPE (%)": wape
    })

results_df = pd.DataFrame(results)

print("MODEL PERFORMANCE COMPARISON")
print(results_df.round(2).to_string(index=False))

# 5. Select the best model using WAPE
valid_results = results_df.dropna(subset=["WAPE (%)"])

if valid_results.empty:
    raise ValueError("Unable to select a model: WAPE is unavailable.")

best_model_name = valid_results.loc[
    valid_results["WAPE (%)"].idxmin(), "Model"
]

print("\nBest Model:", best_model_name)

# 6. Refit the selected model on all historical data
if best_model_name == "Seasonal Naive":

    future_forecast = monthly_revenue.iloc[-12:-6].tolist()

else:
    best_model = ExponentialSmoothing(
        monthly_revenue,
        trend="add",
        damped_trend=(
            best_model_name == "Holt-Winters Damped"
        ),
        seasonal="add",
        seasonal_periods=12,
        initialization_method="estimated"
    ).fit(optimized=True)

    future_forecast = best_model.forecast(6).to_numpy()

# 7. Forecast the next six months
future_dates = pd.date_range(
    start=monthly_revenue.index[-1] + pd.DateOffset(months=1),
    periods=6,
    freq="MS"
)

forecast_df = pd.DataFrame({
    "Month": future_dates.strftime("%B %Y"),
    "Predicted Revenue": future_forecast
})

print("\nNEXT 6 MONTHS REVENUE FORECAST")
print(forecast_df.round(2).to_string(index=False))

print(
    "\nTotal Predicted Revenue for 6 Months:",
    round(np.sum(future_forecast), 2)
)

# 8. Visualize historical and forecast revenue
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_revenue.index,
    monthly_revenue.values,
    label="Historical Revenue"
)

plt.plot(
    future_dates,
    future_forecast,
    marker="o",
    linestyle="--",
    label=f"6-Month Forecast ({best_model_name})"
)

plt.title("Global Sales: Revenue Forecast for Next 6 Months")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

I**nterpretation** – Revenue Forecast for the Next 6 Months
The Holt-Winters Damped model performed best among the three tested models, achieving the lowest WAPE of 43.40%.
The predicted total revenue for the next six months (August 2017 to January 2018) is approximately 75.09 million.
Revenue is forecast to decline from August through December 2017, followed by an increase in January 2018.
Overall Interpretation: The forecast indicates fluctuating revenue, with a potential recovery in January 2018. However, the WAPE of 43.40% indicates substantial forecast errors, so these predictions should be treated as estimates rather than guaranteed results.

```python
#Which region is expected to generate the highest revenue over the next six months based on historical revenue trends?

from statsmodels.tsa.holtwinters import ExponentialSmoothing

# 1. Prepare data
df["Order Date"] = pd.to_datetime(df["Order Date"])

# 2. Calculate monthly revenue for each region
monthly_region_revenue = (
    df.groupby([
        pd.Grouper(key="Order Date", freq="MS"),
        "Region"
    ])["Total Revenue"]
    .sum()
    .unstack("Region", fill_value=0)
    .sort_index()
)

# Fill any missing months with zero revenue
monthly_region_revenue = monthly_region_revenue.asfreq(
    "MS", fill_value=0
)

# 3. Forecast the next 6 months for each region
forecast_results = {}

for region in monthly_region_revenue.columns:

    revenue = monthly_region_revenue[region]

    if len(revenue) < 24:
        print(f"Skipping {region}: insufficient monthly history")
        continue

    model = ExponentialSmoothing(
        revenue,
        trend="add",
        damped_trend=True,
        seasonal="add",
        seasonal_periods=12,
        initialization_method="estimated"
    ).fit(optimized=True)

    forecast_results[region] = model.forecast(6).values

if not forecast_results:
    raise ValueError("No region has enough historical data to forecast.")

# 4. Display six-month forecast by region
future_dates = pd.date_range(
    start=monthly_region_revenue.index[-1] + pd.DateOffset(months=1),
    periods=6,
    freq="MS"
)

forecast_df = pd.DataFrame(
    forecast_results,
    index=future_dates
)
forecast_df.index.name = "Month"

print("REGION-WISE REVENUE FORECAST")
print(forecast_df.round(2))

# 5. Calculate total forecast revenue for each region
region_totals = forecast_df.sum().sort_values(ascending=False)

print("\nTOTAL PREDICTED REVENUE BY REGION")
print(region_totals.round(2))

highest_region = region_totals.idxmax()

print("\nHighest Expected Revenue Region:", highest_region)
print("Predicted Revenue for 6 Months:",
      round(region_totals.max(), 2))

# 6. Visualize the six-month forecast comparison
region_totals.sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Predicted Revenue by Region - Next 6 Months")
plt.xlabel("Predicted Total Revenue")
plt.ylabel("Region")
plt.tight_layout()
plt.show()
```

## 4. Predictive Analysis — What Might Happen?

### Forecasting Methodology

The project uses Holt-Winters Exponential Smoothing, which can model the level, trend, and seasonal patterns of a time series.

The forecasting workflow consists of the following steps:

1. Convert Order Date into a datetime format.
2. Aggregate Total Revenue into monthly totals.
3. Arrange monthly revenue observations in chronological order.
4. Check for missing monthly observations and ensure that the historical series is suitable for forecasting.
5. Split the data into 80% training data and 20% testing data.
6. Fit and compare three forecasting approaches.
7. Evaluate predictions using MAE, RMSE, MAPE, and WAPE.
8. Select the model with the lowest WAPE among the tested models.
9. Refit the selected Holt-Winters model using all available historical data.
10. Forecast revenue for the next six months and visualize the results.

### Forecasting Models Tested

**1. Seasonal Naive Forecast**

This baseline uses revenue from corresponding months in the previous year. It provides a simple benchmark against which more advanced models can be compared.

**2. Holt-Winters Additive**

This model represents the revenue series using an additive trend and additive seasonality. It is suitable when seasonal changes are reasonably consistent in absolute size.

**3. Holt-Winters Damped**

This model uses an additive trend with a damped trend component and additive seasonality. Damping gradually reduces the influence of the projected trend further into the future.

### Understanding the Evaluation Metrics

- **MAE (Mean Absolute Error):** Measures the average absolute difference between actual and predicted revenue. Lower values indicate smaller average errors.
- **RMSE (Root Mean Squared Error):** Penalizes larger prediction errors more heavily. Lower values indicate better performance.
- **MAPE (Mean Absolute Percentage Error):** Measures average absolute percentage error. It can become unstable when actual values are zero or very small.
- **WAPE (Weighted Absolute Percentage Error):** Divides the sum of absolute forecast errors by the sum of actual values. Lower values indicate better performance.

The selected model's WAPE is 43.40%, meaning that the sum of absolute prediction errors is approximately 43.40% of the total actual revenue across the testing period.

Although Holt-Winters Damped performed best among the tested models, its MAPE of 90.10% and WAPE of 43.40% indicate substantial forecast errors. The forecast should therefore be treated as an estimate rather than a highly accurate prediction.
